```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
config_path = Path('../config/nature.mplstyle').resolve()
plt.style.use(str(config_path))

import matplotlib as mpl
mpl.rcParams["figure.dpi"] = 400
```

```python
df1 = pd.read_csv("../data_for_plotting/mercator_cpu_vs_gpu.csv")
df1 = df1[df1['sample_count'] == 'exact']
df1
```

```python
df2 = pd.read_csv("../data_for_plotting/mercator_gpu_vs_negative_sampling.csv")
df2
```

```python
from matplotlib.ticker import FixedLocator, FixedFormatter, NullLocator, NullFormatter

def sci_label(x, pos=None):
    exponent = int(np.log10(x))
    coefficient = x / 10**exponent
    return rf"$\mathregular{{{coefficient:.0f}\times10^{{{exponent}}}}}$"
```

```python
color_map = {'cpu': '#a00e00', 'gpu': '#f6c200'}
color_map2 = {'exact': '#b1615c', '64': '#8282aa'}
label_map = {'exact': 'Exact', '64': 'S=64'}

fig, axes = plt.subplots(1, 3, figsize=(5.5, 1.8))

# CPU vs GPU
for name, group in df1.groupby('backend'):
    axes[0].plot(group['size'], group['total_time_ms']/1000, marker='o', color=color_map[name], label=name.upper())

axes[0].legend()
axes[0].set_xscale('log')
axes[0].set_yscale('log')
axes[0].set_xlabel('Network size')
axes[0].set_ylabel('Time [s]')


# GPU: negative sampling
for sample_count in ['exact', '64']:
    df_tmp = df2[df2['sample_count'] == sample_count]
    axes[1].plot(df_tmp['size'], df_tmp['total_time_ms']/1000, marker='o', color=color_map2[sample_count], label=label_map[sample_count])

axes[1].legend()
axes[1].set_xscale('log')
axes[1].set_yscale('log')
axes[1].set_xlabel('Network size')
axes[1].set_ylabel('Time [s]')

ticks = [50000, 100000, 500000]
axes[1].xaxis.set_major_locator(FixedLocator(ticks))
axes[1].xaxis.set_major_formatter(FixedFormatter([sci_label(x) for x in ticks]))
axes[1].xaxis.set_minor_locator(NullLocator())

# GPU: negative sampling
for sample_count in ['exact', '64']:
    df_tmp = df2[df2['sample_count'] == sample_count]
    axes[2].plot(df_tmp['size'], df_tmp['c_score'], marker='o', color=color_map2[sample_count], label=label_map[sample_count])

axes[2].legend()
axes[2].set_xscale('log')
axes[2].set_xlabel('Network size')
axes[2].set_ylabel('C-score')

ticks = [50000, 100000, 500000]
axes[2].xaxis.set_major_locator(FixedLocator(ticks))
axes[2].xaxis.set_major_formatter(FixedFormatter([sci_label(x) for x in ticks]))
axes[2].xaxis.set_minor_locator(NullLocator())

panel_letters = ['a', 'b', 'c']
for ax, letter in zip(axes, panel_letters):
    ax.text(
        -0.17,
        1.12,
        letter,
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=10,
    )

plt.tight_layout()
# plt.savefig("mercator_time_complexity.pdf")
```
